```r
# The first two arguments to ggplot() are data and mapping

library(tidyverse)
# install.packages("palmerpenguins")
library(palmerpenguins)
```

## 1.4 Visualizing distributions

### 1.4.1 A categorical variable

```r
ggplot(penguins, aes(x = species)) +
  geom_bar() # bar chart
```

```r
ggplot(penguins, aes(x = fct_infreq(species))) + # reorder the bars based on their frequencies
  geom_bar()
  # factor infrequent
```

### 1.4.2 A numerical variable

```r
ggplot(penguins, aes(x = body_mass_g)) +
  geom_histogram(binwidth = 200)
```

```r
ggplot(penguins, aes(x = body_mass_g)) +
  geom_histogram(binwidth = 20)
ggplot(penguins, aes(x = body_mass_g)) +
  geom_histogram(binwidth = 2000)
```

```r
ggplot(penguins, aes(x = body_mass_g)) +
  geom_density() # smoothed-out version of a histogram
```
